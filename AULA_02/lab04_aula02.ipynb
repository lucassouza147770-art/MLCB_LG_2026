```python
# ==============================================================================
# PROTÓTIPO DE MOTOR NLU - CHATBOT AGÊNCIA DE VIAGENS
# ==============================================================================

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 1. Criação do Dataset Próprio (Mínimo 12 frases e 3 intenções)
dados_viagem = {
    'mensagem': [
        # Intenção: comprar_passagem
        'Quero comprar uma passagem para Paris',
        'Como faço para reservar um voo para o Rio de Janeiro?',
        'Preciso de uma passagem aérea barata para Miami',
        'Gostaria de emitir um bilhete de avião para Lisboa',

        # Intenção: cancelar_reserva
        'Preciso cancelar minha reserva de hotel',
        'Quero desistir da minha viagem e pedir reembolso',
        'Como faço para fazer o cancelamento do meu voo?',
        'Gostaria de estornar a passagem que comprei ontem',

        # Intenção: falar_atendente
        'Quero falar com um atendente humano',
        'Pode me transferir para o suporte especializado?',
        'Preciso de ajuda de uma pessoa de verdade',
        'Chame um atendente por favor'
    ],
    'intencao': [
        'comprar_passagem', 'comprar_passagem', 'comprar_passagem', 'comprar_passagem',
        'cancelar_reserva', 'cancelar_reserva', 'cancelar_reserva', 'cancelar_reserva',
        'falar_atendente', 'falar_atendente', 'falar_atendente', 'falar_atendente'
    ]
}

df_viagem = pd.DataFrame(dados_viagem)

# Separação de Atributos (X) e Rótulos (y)
X = df_viagem['mensagem']
y = df_viagem['intencao']


# 2. Divisão do Dataset em Treino e Teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)


# 3. Escolha do Vetorizador de Texto (TfidfVectorizer)
vectorizer = TfidfVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)


# 4. Escolha do Algoritmo de Classificação e Treinamento (LogisticRegression)
modelo_nlu = LogisticRegression()
modelo_nlu.fit(X_train_vec, y_train)


# Avaliação opcional do modelo com o conjunto de teste
y_pred_test = modelo_nlu.predict(X_test_vec)
acuracia = accuracy_score(y_test, y_pred_test)


# 5. Predições com no mínimo 3 frases INÉDITAS
frases_ineditas = [
    "Quero emitir um voo para o Japão na próxima semana",
    "Não vou mais viajar e quero meu dinheiro de volta",
    "Pode me passar para o atendimento humano?"
]

# Transforma as novas frases usando o mesmo vetorizador
frases_ineditas_vec = vectorizer.transform(frases_ineditas)

# Gera as predições reais das frases inéditas
predicoes_finais = modelo_nlu.predict(frases_ineditas_vec)


# 6. Exibição das intenções previstas no console
print("==================================================")
print("     RESULTADOS DO MOTOR NLU - AGÊNCIA DE VIAGENS  ")
print("==================================================")
print(f"Acurácia interna no teste: {acuracia * 100:.2f}%\n")
print("--- Testando Frases Inéditas dos Usuários ---")

for frase, intencao in zip(frases_ineditas, predicoes_finais):
    print(f"Usuário: '{frase}'")
    print(f"🤖 Intenção Detectada: [{intencao}]")
    print("-" * 50)
```
