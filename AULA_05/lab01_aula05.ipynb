```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import nltk
import spacy
from nltk.corpus import stopwords
from gensim.models import FastText
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

nltk.download("stopwords", quiet=True)
stop_words_pt = set(stopwords.words("portuguese"))
nlp = spacy.load("pt_core_news_sm")
print("Ambiente preparado com sucesso.")

def limpar_e_lemmatizar(texto):
    """
    Realiza o pré-processamento de uma mensagem:

    1. Converte para minúsculas
    2. Remove caracteres especiais e números
    3. Processa o texto com spaCy
    4. Remove stopwords
    5. Realiza lemmatization
    6. Retorna o texto normalizado
    """

    # 1. Converter para minúsculas
    texto_limpo = texto.lower()

    # 2. Remover caracteres especiais e números
    texto_limpo = re.sub(
        r'[^a-záàâãéèêíïóôõöúçñ\s]',
        '',
        texto_limpo
    )

    # 3. Processar o texto com spaCy (Separa as palavras ex: "oi meu nome é..."  result:"oi", "meu", "nome", "é", "...")
    doc = nlp(texto_limpo)

    # 4. Extrair os lemas (transforma na palavra original de dicionário.)
    tokens_filtrados = []

    for token in doc:

        # Ignorar espaços
        if token.is_space:
            continue

        # Ignorar stopwords
        if token.text in stop_words_pt:
            continue

        # Ignorar tokens muito pequenos
        if len(token.text) <= 1:
            continue

        # Adicionar o lema
        tokens_filtrados.append(token.lemma_)

    # 5. Reconstruir a frase
    return " ".join(tokens_filtrados)
```

```python
# @title Atividade 1
frase_teste = (
   "MEU sofá!!! chegou quebrado e quero DEVOLVER!!!"
)

resultado = limpar_e_lemmatizar(frase_teste)

print("Frase original:")
print(frase_teste)

print("\nFrase processada:")
print(resultado)
```

```python
# ==========================================
# 1. ADAPTAÇÃO DA FUNÇÃO PARA RETORNAR LISTA
# ==========================================
def limpar_e_lemmatizar_lista(texto):
    """
    Recebe uma frase e devolve uma LISTA de lemas limpos.
    O FastText precisa exatamente desse formato (lista de palavras).
    """
    if pd.isna(texto):
        return []

    texto_limpo = str(texto).lower()
    texto_limpo = re.sub(r'[^a-záàâãéèêíïóôõöúçñ\s]', '', texto_limpo)
    doc = nlp(texto_limpo)

    tokens_filtrados = []
    for token in doc:
        if token.is_space or token.text in stop_words_pt or len(token.text) <= 1:
            continue
        tokens_filtrados.append(token.lemma_)

    return tokens_filtrados

# ==========================================
# 2. CARREGANDO O ARQUIVO DE TREINO DO COLAB
# ==========================================
# Mudamos o nome do arquivo aqui para bater com o que está na sua barra lateral
df_treino = pd.read_csv("sac_moveis_ac2_treino.csv")

print("Normalizando as mensagens de treino...")
# Cria a coluna com as palavras separadas em lista
df_treino['mensagem_tokenizada'] = df_treino['mensagem'].apply(limpar_e_lemmatizar_lista)

# Extrai o corpus tokenizado exigido pelo FastText
corpus_tokenizado = df_treino['mensagem_tokenizada'].tolist()

# ==========================================
# 3. TREINANDO O MODELO FASTTEXT
# ==========================================
print("Treinando o modelo FastText...")
modelo_fasttext = FastText(
    sentences=corpus_tokenizado,
    vector_size=50,
    window=3,
    min_count=1,
    workers=4,
    sg=1
)
print("Modelo FastText treinado com sucesso!")
```

```python
def obter_vetor_frase(frase, modelo):
    """
    Transforma uma frase em um vetor denso utilizando
    FastText + Mean Pooling.
    """

    palavras = frase.split()

    vetores = []

    for palavra in palavras:

        # TODO 1:
        # Obter o vetor da palavra no modelo FastText
        # No Gensim 4.0+, os vetores ficam dentro de .wv (word vectors)
        if palavra in modelo.wv:
            vetor = modelo.wv[palavra]

            # TODO 2:
            # Adicionar o vetor à lista
            vetores.append(vetor)

    # TODO 3:
    # Verificar se nenhum vetor foi encontrado
    # Se a lista estiver vazia, retorna um vetor de zeros com a dimensão do modelo (50)
    if len(vetores) == 0:
        return np.zeros(modelo.vector_size)

    # TODO 4:
    # Calcular o Mean Pooling
    # O np.mean calcula a média matemática elemento por elemento
    vetor_medio = np.mean(vetores, axis=0)
    return vetor_medio

```

```python
# @title Atividade 2
frase_teste = (
   "MEU sofá!!! chegou quebrado e quero DEVOLVER!!!"
)

# 1. Limpa e extrai os lemas da frase original
resultado_1 = limpar_e_lemmatizar(frase_teste)

# 2. Transforma o texto limpo em vetor (passando o modelo_fasttext como segundo argumento!)
resultado = obter_vetor_frase(resultado_1, modelo_fasttext)

print("Frase original:")
print(frase_teste)

print("\nFrase processada (Texto limpo):")
print(resultado_1)

print("\nRepresentação Semântica (Vetor de 50 dimensões gerado pelo Mean Pooling):")
print(resultado)
```
